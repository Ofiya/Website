```python
# Create folders for Planner data in the default Lakehouse Files area
# This code requires Microsoft Fabric Runtime kernel to work

base_path = "Files"

folders = [
    "PlannerTasks",
    "PlannerPlans",
    "PlannerBuckets"
]

for folder in folders:
    try:
        # Use mssparkutils (available in Fabric Runtime kernel)
        mssparkutils.fs.mkdirs(f"{base_path}/{folder}")
        print(f"Created (or already exists): {base_path}/{folder}")
    except Exception as e:
        print(f"Failed to create {base_path}/{folder}: {e}")
```

```python

# Create Delta tables mapped to Tables/ folder locations

spark.sql("""
CREATE TABLE IF NOT EXISTS PlannerTasks (
    taskId STRING,
    planId STRING,
    bucketId STRING,
    title STRING,
    assignedTo ARRAY<STRING>,
    dueDate TIMESTAMP,
    percentComplete INT,
    createdDateTime TIMESTAMP,
    lastModifiedDateTime TIMESTAMP
)
USING DELTA
""")
print("Created table: PlannerTasks")

spark.sql("""
CREATE TABLE IF NOT EXISTS PlannerPlans (
    planId STRING,
    title STRING,
    owner STRING,
    createdDateTime TIMESTAMP
)
USING DELTA
""")
print("Created table: PlannerPlans")

spark.sql("""
CREATE TABLE IF NOT EXISTS PlannerBuckets (
    bucketId STRING,
    planId STRING,
    name STRING,
    orderHint STRING
)
USING DELTA
""")
print("Created table: PlannerBuckets")
```
